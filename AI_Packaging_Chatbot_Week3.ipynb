```python
questions = [item["question"] for item in knowledge_base]

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english"
)

question_vectors = vectorizer.fit_transform(questions)

print("AI retrieval system is ready!")
```

```python
def chatbot_response(user_input, threshold=0.25):

    user_input = user_input.strip()

    if not user_input:
        return "Please enter a question so I can help you."

    user_vector = vectorizer.transform([user_input])

    similarity_scores = cosine_similarity(
        user_vector,
        question_vectors
    )[0]

    best_index = similarity_scores.argmax()
    best_score = similarity_scores[best_index]

    if best_score < threshold:
        return (
            "I'm sorry, I don't have enough information in my knowledge base "
            "to answer that accurately. Please contact our sales team for assistance."
        )

    return knowledge_base[best_index]["answer"]
```

```python
test_questions = [
    "What is your minimum order?",
    "Can you make custom boxes?",
    "What materials do you use?",
    "How long does production take?",
    "Can I print my logo?"
]

for question in test_questions:
    print("Customer:", question)
    print("Chatbot:", chatbot_response(question))
    print("-" * 60)
```

```python
def chatbot_with_score(user_input, threshold=0.25):

    user_input = user_input.strip()

    if not user_input:
        return "Please enter a question.", 0

    user_vector = vectorizer.transform([user_input])

    similarity_scores = cosine_similarity(
        user_vector,
        question_vectors
    )[0]

    best_index = similarity_scores.argmax()
    best_score = similarity_scores[best_index]

    if best_score < threshold:
        return (
            "I'm sorry, I don't have enough information to answer that accurately.",
            best_score
        )

    return knowledge_base[best_index]["answer"], best_score
```

```python
answer, score = chatbot_with_score("Do you make kraft boxes?")

print("Answer:", answer)
print("Confidence Score:", round(score, 3))
```

```python
def collect_quote():

    print("\n--- Packaging Quote Request ---")

    product = input("Product type: ")
    length = input("Length: ")
    width = input("Width: ")
    height = input("Height: ")
    quantity = input("Quantity: ")
    material = input("Material: ")
    printing = input("Printing requirements: ")
    finishing = input("Finishing: ")
    city = input("Delivery city: ")

    quote = {
        "product": product,
        "dimensions": f"{length} x {width} x {height}",
        "quantity": quantity,
        "material": material,
        "printing": printing,
        "finishing": finishing,
        "delivery_city": city,
        "date": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }

    print("\nQuote information collected successfully!")
    print(json.dumps(quote, indent=4))

    return quote
```

```python
quote_data = collect_quote()
```

```python
with open("quote_requests.json", "a", encoding="utf-8") as file:
    json.dump(quote_data, file, indent=4)
    file.write("\n")

print("Quote request saved successfully!")
```

```python
def run_chatbot():

    print("\n" + "=" * 60)
    print("WELCOME TO PACKPRO SOLUTIONS")
    print("AI-Powered Packaging Assistant")
    print("=" * 60)

    print("Ask me about products, materials, MOQ, samples,")
    print("production, delivery and quotations.")
    print("Type 'quote' for a quotation request.")
    print("Type 'exit' to close the chatbot.")

    while True:

        user_input = input("\nYou: ").strip()

        if user_input.lower() == "exit":
            print("Chatbot: Thank you for contacting PackPro Solutions!")
            break

        if user_input.lower() == "quote":
            collect_quote()
            continue

        response, score = chatbot_with_score(user_input)

        print("\nChatbot:", response)

        if score > 0:
            print("(Retrieval confidence:", round(score, 3), ")")
```

```python
run_chatbot()
```

```python
conversation_log = []

def log_conversation(user_input, response):

    record = {
        "time": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "user": user_input,
        "chatbot": response
    }

    conversation_log.append(record)

    with open("conversation_log.json", "w", encoding="utf-8") as file:
        json.dump(conversation_log, file, indent=4)
```
